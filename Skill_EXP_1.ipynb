```python
import pandas as pd

# ==========================
# Load Dataset
# ==========================
CSV_PATH = "/content/nifty_500.csv"

df = pd.read_csv(CSV_PATH)

# ==========================
# Convert text columns to numeric
# ==========================
numeric_text_columns = [
    "Change",
    "Percentage Change",
    "365 Day Percentage Change",
    "30 Day Percentage Change"
]

for col in numeric_text_columns:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace("%", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip()
    )
    df[col] = pd.to_numeric(df[col], errors="coerce")

# ==========================
# 1. First 5 rows
# ==========================
print("FIRST 5 ROWS")
print(df.head())

# ==========================
# 2. Dataset Shape
# ==========================
print("\nDATASET SHAPE")
print(df.shape)

# ==========================
# 3. Column Names
# ==========================
print("\nCOLUMN NAMES")
print(df.columns)

# ==========================
# 4. Data Types
# ==========================
print("\nDATA TYPES")
print(df.dtypes)

# ==========================
# 5. Dataset Information
# ==========================
print("\nDATASET INFORMATION")
df.info()

# ==========================
# 6. Statistical Summary
# ==========================
print("\nSTATISTICAL SUMMARY")
print(df.describe(include="all"))

# ==========================
# 7. Missing Values
# ==========================
print("\nMISSING VALUES")
print(df.isnull().sum())

# ==========================
# 8. Duplicate Rows
# ==========================
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

# ==========================
# 9. Industry Distribution
# ==========================
print("\nINDUSTRY DISTRIBUTION")
print(df["Industry"].value_counts())

# ==========================
# 10. Top 10 Companies by Share Volume
# ==========================
print("\nTOP 10 COMPANIES BY SHARE VOLUME")
print(df.nlargest(10, "Share Volume")[["Company Name", "Share Volume"]])

# ==========================
# 11. Top 10 Companies by Value Traded
# ==========================
print("\nTOP 10 COMPANIES BY VALUE TRADED")
print(df.nlargest(10, "Value (Indian Rupee)")[["Company Name", "Value (Indian Rupee)"]])

# ==========================
# 12. Top 10 Gainers
# ==========================
print("\nTOP 10 GAINERS")
print(df.nlargest(10, "Percentage Change")[["Company Name", "Percentage Change"]])

# ==========================
# 13. Top 10 Losers
# ==========================
print("\nTOP 10 LOSERS")
print(df.nsmallest(10, "Percentage Change")[["Company Name", "Percentage Change"]])

# ==========================
# 14. Average Opening Price
# ==========================
print("\nAVERAGE OPENING PRICE")
print(df["Open"].mean())

# ==========================
# 15. Average Last Traded Price
# ==========================
print("\nAVERAGE LAST TRADED PRICE")
print(df["Last Traded Price"].mean())

# ==========================
# 16. Company with Highest 52-Week High
# ==========================
print("\nCOMPANY WITH HIGHEST 52-WEEK HIGH")
highest = df.loc[df["52 Week High"].idxmax()]
print(highest[["Company Name", "52 Week High"]])

# ==========================
# 17. Company with Lowest 52-Week Low
# ==========================
print("\nCOMPANY WITH LOWEST 52-WEEK LOW")
lowest = df.loc[df["52 Week Low"].idxmin()]
print(lowest[["Company Name", "52 Week Low"]])

# ==========================
# 18. Average 365-Day Percentage Change
# ==========================
print("\nAVERAGE 365-DAY PERCENTAGE CHANGE")
print(df["365 Day Percentage Change"].mean())

# ==========================
# 19. Average 30-Day Percentage Change
# ==========================
print("\nAVERAGE 30-DAY PERCENTAGE CHANGE")
print(df["30 Day Percentage Change"].mean())

# ==========================
# 20. Positive vs Negative Daily Returns
# ==========================
print("\nDAILY RETURN DISTRIBUTION")
print(
    df["Percentage Change"]
    .apply(lambda x: "Positive" if x >= 0 else "Negative")
    .value_counts()
)
```
